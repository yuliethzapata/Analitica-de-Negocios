##**Caso de Estudio**
Implementar un modelo de árbol de decisión para llevar a cabo un proceso de pronóstico de si las personas que hacen parte de esta base de datos van a desarrollar una enfermedad cardiaca. Para este proceso vamos a utilizar las variables:
* Age: Indica el número de años que posee una persona, Representa la edad cronológica del paciente.
* Systolic: Presión Sistólica la cual es la presión en las arterias cuando el corazón late (el número "alto"). El árbol la usará para identificar hipertensión, un factor crítico que daña las paredes arteriales.
* Diastolic: Presión Diastólica la cual es la presión en las arterias cuando el corazón descansa entre latidos (el número "bajo"). Ayuda al modelo a evaluar la resistencia vascular periférica y la salud del corazón en reposo.
* BMI (EAD): Índice de Masa Corporal (IMC). Es un cálculo basado en peso y altura (Peso sobre Altura, y la altura va al cuadrado. El modelo lo utiliza como un indicador indirecto de obesidad, la cual está estrechamente ligada a la inflamación y al esfuerzo cardíaco extra.

0. Se procede con la carga de las librerias de trabajo

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix
```

1. Se procede con la carga de datos de trabajo

```python
import pandas as pd
nxl = '/content/2. HeartDisease (1).xlsx'
XDB=pd.read_excel(nxl,sheet_name=1)
XD=XDB.iloc[:,[1,2,3,4]]
display(XD)
```

2. Se procede con la implementación del modelo de árbol

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix
mar=DecisionTreeClassifier(criterion='gini',max_depth=4)
yd = XDB.iloc[:, -1] # Assuming the last column of XDB is the target variable
mar.fit(XD,yd) #Aquí el modelo busca la relación entrada-salida

#¿y que fué lo que hizo el modelo?
ydp=mar.predict(XD) #Esto es lo que pronóstica el modelo

#Se construye la matriz de confusón
cm=confusion_matrix(yd,ydp)
display(cm)
VN=cm[0,0]; FP=cm[0,1]; FN=cm[1,0]; VP=cm[1,1]
#Médtricas de Desempeño
Ex=(VP+VN)/len(XD) #1. Exaltitud: Comportamiento General
__builtins__.print("La exaltitud es:",Ex)
sen=VP/(VP+FN) #2. Sensibilidad: Como se comporta pronósticando PreApr
__builtins__.print("La sensibilidad es:",sen)
Spe=VN/(VN+FP) #3. Especifidad: Como se comporta frente a los negativos
__builtins__.print("La especificidad es:", Spe)
pre=VP/(VP+FP) #4. Precisión: Como se comporta pronósticando PreApr
__builtins__.print("La precisión es:",pre)
PreNeg=VN/(VN+FN) #5. Precisión Negativa: Como se comporta pronósticando negativos
__builtins__.print("La predicción negativa es:",PreNeg)
```

3. Despliegue del Árbol de Decisión

```python
from sklearn.tree import export_graphviz #Exporta los datos a un gráfico
from pydotplus import graph_from_dot_data #Es un graficador

vs=["Age","Systolic","Diastolic","BMI"] #Títulos del árbol
dot_data=export_graphviz(mar,feature_names=vs) #Exportar de números a gráfico en pdf
graph=graph_from_dot_data(dot_data)               #Hacemos el gráfico
graph.write_png('Arbol.png')
```
